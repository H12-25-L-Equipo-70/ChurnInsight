# **Ingeniería de Features**

```python
# ============================================
# 2. Ingeniería de Features
# ============================================

# --- Ratios financieros ---
df['MARGEN_RATIO'] = df['MARGEN'] / df['INGRESOS']
df['DEUDA_ACTIVOS_RATIO'] = df['DEUDA'] / df['ACTIVOS']

# --- Préstamos ---
df['TASA_APROBACION_PRESTAMOS'] = df['PRESTAMOS_APROBADOS'] / df['PRESTAMOS_SOLICITADOS'].replace(0, pd.NA)
df['TASA_APROBACION_MONTO'] = df['MONTO_APROBADO'] / df['MONTO_SOLICITADO'].replace(0, pd.NA)

# --- Actividad digital ---
df['RATIO_DIAS_ACTIVOS'] = df['TRIMESTRE_DIAS_ACTIVIDAD'] / (df['TRIMESTRE_DIAS_ACTIVIDAD'] + df['TRIMESTRE_DIAS_INACTIVIDAD'])
df['ENGAGEMENT_LOGIN'] = df['TOTAL_LOGIN_DIA'] / df['TRIMESTRE_DIAS_ACTIVIDAD'].replace(0, pd.NA)

# --- Historial de pagos/servicios ---
df['USO_SERVICIOS'] = df[['TRANSFERENCIAS','PAGOS','CREDITOS','INVERSIONES']].sum(axis=1)

# --- Periodo fiscal: año y trimestre ---
#df['ANIO'] = df['PERIODO_FISCAL'].str.split('-').str[0].astype(int)
#df['TRIMESTRE'] = df['PERIODO_FISCAL'].str.split('-').str[1]
# Usamos .strip() para eliminar espacios en blanco accidentales
df['ANIO'] = df['PERIODO_FISCAL'].str.split('-').str[0].str.strip().astype(int)
df['TRIMESTRE'] = df['PERIODO_FISCAL'].str.split('-').str[1].str.strip()

# Verificación rápida: ¿Qué valores únicos existen ahora?
print("Trimestres detectados:", df['TRIMESTRE'].unique())
print("Conteo por trimestre:\n", df['TRIMESTRE'].value_counts())
# One-hot encoding de trimestre1
df = pd.get_dummies(df, columns=['TRIMESTRE'], drop_first=False)

print("\nNuevas columnas creadas:")
print([col for col in df.columns if col not in ['NOMBRE_EMPRESA','CUIT','PERIODO_FISCAL']])
```

```python
from sklearn.model_selection import train_test_split

X = df.drop(columns=['CHURN','NOMBRE_EMPRESA','CUIT','PERIODO_FISCAL'])
y = df['CHURN']

X_train, X_test, y_train, y_test = train_test_split(
    X, y, test_size=0.3, random_state=42, stratify=y
)
```

```python
print("Distribución en train:", y_train.value_counts(normalize=True))
print("Distribución en test:", y_test.value_counts(normalize=True))
```

```python
from sklearn.model_selection import train_test_split
from sklearn.ensemble import RandomForestClassifier
from sklearn.metrics import classification_report, confusion_matrix, roc_auc_score, RocCurveDisplay
import numpy as np # Import numpy for np.nan
from imblearn.over_sampling import SMOTE
from sklearn.impute import SimpleImputer # Import SimpleImputer

# Convert pd.NA values to np.nan in X_train and X_test
X_train = X_train.replace(pd.NA, np.nan)
X_test = X_test.replace(pd.NA, np.nan)

# Impute missing values before SMOTE
imputer = SimpleImputer(strategy='mean')
X_train = pd.DataFrame(imputer.fit_transform(X_train), columns=X_train.columns)
X_test = pd.DataFrame(imputer.transform(X_test), columns=X_test.columns)

smote = SMOTE(random_state=42)
X_train_bal, y_train_bal = smote.fit_resample(X_train, y_train)

model = RandomForestClassifier(
     n_estimators=200, max_depth=None, random_state=42, class_weight=None
     # ya balanceamos con SMOTE
     )
model.fit(X_train_bal, y_train_bal)
```

```python
y_pred = model.predict(X_test)
y_proba = model.predict_proba(X_test)[:,1]
print("\nReporte de clasificación:")
print(classification_report(y_test, y_pred))
print("\nMatriz de confusión:")
print(confusion_matrix(y_test, y_pred))
print("\nROC-AUC:", roc_auc_score(y_test, y_proba))
```

```python
# Curva ROC
RocCurveDisplay.from_predictions(y_test, y_proba)
plt.show()
```

```python
importances = pd.Series(model.feature_importances_, index=X.columns).sort_values(ascending=False)
plt.figure(figsize=(10,6))
sns.barplot(x=importances[:15], y=importances.index[:15])
plt.title("Top 15 Features más importantes")
plt.show()
```

```python
#XGboost
#Entrenamiento
from xgboost import XGBClassifier
model_xgb = XGBClassifier( n_estimators=300, learning_rate=0.05, max_depth=6, subsample=0.8, colsample_bytree=0.8, random_state=42, use_label_encoder=False, eval_metric='logloss' )
model_xgb.fit(X_train_bal, y_train_bal)

```

```python
# Evaluación con XGboost
y_pred_xgb = model_xgb.predict(X_test)
y_proba_xgb = model_xgb.predict_proba(X_test)[:,1]

print("\nReporte de clasificación (XGBoost):")
print(classification_report(y_test, y_pred_xgb))
print("\nMatriz de confusión (XGBoost):")
print(confusion_matrix(y_test, y_pred_xgb))
```

```python
print("\nROC-AUC (XGBoost):", roc_auc_score(y_test, y_proba_xgb)) # Curva ROC
RocCurveDisplay.from_predictions(y_test, y_proba_xgb)
plt.show()
```

```python
# Features importantes
importances_xgb = pd.Series(model_xgb.feature_importances_, index=X.columns).sort_values(ascending=False)
plt.figure(figsize=(10,6))
sns.barplot(x=importances_xgb[:15], y=importances_xgb.index[:15])
plt.title("Top 15 Features más importantes (XGBoost)")
plt.show()
```

```python
import matplotlib.pyplot as plt
import seaborn as sns
import numpy as np

# 1. Configuración de lógica de negocio Fintech
# Excluimos CUIT y variables que no son predictoras
columnas_a_excluir = ['CUIT']
if 'ID_CLIENTE' in df.columns: columnas_a_excluir.append('ID_CLIENTE')

# 2. Cálculo de correlaciones
matriz_corr = df.drop(columns=columnas_a_excluir).corr(numeric_only=True)
# Extraemos la relación con CHURN y eliminamos la fila CHURN consigo misma
correlaciones_churn = matriz_corr['CHURN'].drop('CHURN').sort_values(ascending=False)

# 3. Visualización Profesional
plt.figure(figsize=(12, 8))

# Creamos una paleta condicional: Rojo si aumenta el Churn, Verde si lo previene
colores = ['#e74c3c' if x > 0 else '#2ecc71' for x in correlaciones_churn.values]

ax = sns.barplot(x=correlaciones_churn.values,
                 y=correlaciones_churn.index,
                 palette=colores)

# Añadimos los valores numéricos al final de cada barra para precisión
ax.bar_label(ax.containers[0], fmt='%.2f', padding=5)

# Estética del gráfico
plt.title("Factores de Riesgo y Retención (Correlación vs CHURN)", fontsize=15, pad=20)
plt.xlabel("Grado de Influencia (Coeficiente de Pearson)", fontsize=12)
plt.ylabel("Variables de la Fintech", fontsize=12)
plt.axvline(x=0, color='black', linestyle='-', linewidth=1) # Eje central
plt.grid(axis='x', linestyle='--', alpha=0.6) # Rejilla sutil para lectura

# Añadir una nota explicativa
plt.figtext(0.5, -0.05,
            "Interpretación: Barras ROJAS indican factores que impulsan la fuga.\n"
            "Barras VERDES indican factores que ayudan a retener al cliente.",
            ha="center", fontsize=10, bbox={"facecolor":"orange", "alpha":0.1, "pad":5})

plt.show()
```
